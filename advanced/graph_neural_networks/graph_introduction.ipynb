```python
import networkx as nx
```

```python
my_nodes = list(range(5))
H = nx.DiGraph()
```

```python
H.add_nodes_from(my_nodes)

# Add edges
H.add_edges_from([(0, 1), (0, 2), (1, 2), (2, 0), (2, 3), (3, 3), (4, 1)])
```

```python
nx.draw(H, with_labels=True)
```

```python
# Get dense adjacenty matrix
nx.adjacency_matrix(H).todense()
```
